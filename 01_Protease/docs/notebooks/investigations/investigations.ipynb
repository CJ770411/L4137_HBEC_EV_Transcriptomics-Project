# i-120526-01 - cutadapt (trimming)

**Description**
Investigation into the effect of adjusting raw read trimming parameters.

**Tests**
- Test 1 = TGGAATTCTCGG adapter
- Test 2 = TGGAATTCTCGGGTGCCAAGG adapter
- Test 3 = TGGAATTCTCGGGTGCCAAGG adapter + no min quality
- Test 4 = TGGAATTCTCGGGTGCCAAGG adapter + overlap 
- Test 5 = TGGAATTCTCGGGTGCCAAGG adapter + only min read length (17nt)
- Test 6 = TGGAATTCTCGGGTGCCAAGG adapter + only min read length (17nt) + quality filter (q20)

**Conclusions**
1. No difference between test 1 and test 2; TGGAATTCTCGGGTGCCAAGG adapter targeting selected as final parameter because it is more specific.
2. No difference between test 2 and test 3; quality filter removed as parameter to increase simplicity
3. Identical results between test 3 and test 4; overlap parameter excluded from command to increase simplicity.
4. Massive difference in sequence quality histogram between test 4 and test 5; removing max read length parameter makes big difference - further testing required (see below)
5. Little difference between test 5 and test 6, the quality filter didn't massively improve sequence quality histogram results; going to include max read length during further tests.

**Outcome**
- It is unclear whether a maximum read length should be defined during trimming. Some sources say it is necessary ([ref1](https://link.springer.com/protocol/10.1007/978-1-0716-2067-0_26)), [ref2](https://bioinfo-dirty-jobs.github.io/NGSBASE//lectures/05.quality_control/). However, other sources do not define a maximum such as [TamiRNA](https://www.mdpi.com/1422-0067/23/3/1226) or [Huang et al., 2025](https://www.jove.com/t/68760/a-verified-workflow-for-mirna-seq-data-processing-bioinformatics).
- Therefore, a subsequent investigation will run the pipeline with/without max length filtering and compare the results.
- On review, the majority of documentation for miRNA diff expression analysis use 18nt min length instead of 17; min length will be adjusted to 18nt to align with this.
- Final command:

```{bash}
cutadapt \
    -j "$SLURM_CPUS_PER_TASK" \
    -a TGGAATTCTCGGGTGCCAAGG \
    -m 18 \
    -M 30 \
    -o "${OUTDIR_CUTADAPT}/${SAMPLE_NAME}_trimmed.fastq.gz" \
    "$sample"
```



#### Test 1
**Description**
- Adapter targeted for removal is TGGAATTCTCGG
- This is Illumina Small RNA 3' adapter ([reference](https://bioinformatics.ccr.cancer.gov/docs/b4b/10.-Removing-adapters-from-sequence-data%20copy/)) which was present in 20/20 overrepresented sequences from the raw multiqc report.

**Command**

```{bash}
cutadapt \
    -j "$SLURM_CPUS_PER_TASK" \
    -a TGGAATTCTCGG \
    -m 17 \
    -M 30 \
    -q 20 \
    -o "${OUTDIR_CUTADAPT}/${SAMPLE_NAME}_trimmed.fastq.gz" \
    "$sample"
```



**Results**
Negligible difference between test 5 and test 6 'sequence quality histogram'. TamiRNA's histogram shows much stronger quality.



# i-210526-01 - miRDeep2 (count quantification)

**Description**
An initial execution of quantifier.pl from miRDeep2 was performed, and the raw read counts were compared to TamiRNA, producing results that fell into one of three categories (more detail on this in `project_worklog.ipynb`):

1. **Double** - CJ miRNA counts were approximately double TamiRNA counts.
2. **Identical** - CJ and TamiRNA miRNA counts were approximately (or exactly) identical.
3. **Different** - CJ and TamiRNA miRNA counts were vastly different.

- The 'double' category was attributed to miRNAs derived from multiple precursors being aggregated in CJ analysis but not by TamiRNA. 
- The 'identical' category was reassuring as it shows that where multiple precursors didn't exist, it was possible to get the same results.
- The 'different' category was concerning as in some cases CJ found 0 reads and TamiRNA found >10, or vice versa. Indicates that some reads were potentially excluded in upstream analysis. Quick visual inspection showed this category was the least prominent.

- This investigation will alter the parameters of quantifier.pl from miRDeep2 to see if they increase concordance of results with TamiRNA. 
- However, reads derived from multiple precursors still contribute to the overall miRNA abundance therefore are still crucial in determining differential expression and CJ method of aggregation may still be adopted.
- Read quantification using bowtie + featurecounts/htseq is still outstanding therefore results from that analysis must be considered in addition to this investigation. CJ 21.05.2026

**Tests**

- Test 1: Original parameters used during initial execution of quantifier.pl.
- Test 2: Inclusion of `-U` flag to only consider uniquely mapping reads (discards all multimappers)
- Test 3: Inclusion of `-W` flag to weigh reads by the number of mappings
- Test 4: Inclusion of `-x` to skip mapping against precursors
- Test 5: Observe whether the counts would be more comparable if only counts from the precursor with the highest read count is retained; could then produce simple script to retain only these counts. **Test remains outstanding. CJ 21.05.2026**

**Conclusions**

Results acquired by observing `NGS-110-030_S30_R1_001`.

1. Test 1 = Combination of results: 1. miRNA counts approx. double the count of TamiRNA; 2. miRNA counts identical (or approximately identical) to TamiRNA; 3. miRNA counts markedly dissimilar to TamiRNA.
2. Test 2 = Large reduction in reads, making some counts that were similar to TamiRNA's more different than without the `-U` flag. Don't think TamiRNA used this flag in their analysis.
3. Test 3 = Creates floating number raw counts which are incompatible with DE analysis software, which require integers.
4. Test 4 = Didn't run. Mapping is required for quantifier.pl to work.
5. Test 5 = **Outstanding. CJ 21.05.2026**


**Outcome**
- Pending the outcome of **Test 5**, the initial script appears to be most suitable.
- Will repeat read count quantification using bowtie + featurecounts/htseq and compare the results. CJ 21.05.2026
- Final command:

```{bash}

quantifier.pl \
-p "$INFILE_HAIR_MIRNA" \
-m "$INFILE_MAT_MIRNA" \
-r "$INFILE_COLLAPSED_READS" \
-t Human

```
