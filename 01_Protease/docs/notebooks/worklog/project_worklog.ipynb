**5. Commands**
mapper.pl - mapping (collapsed reads)

- `-s` = processed reads printed to this file (only one instance of a read is written accompanied by the number of times it appears)
- `-t` = read mappings printed to this file (alignment)
- `-l 18` = discard reads shorted than 18nt
- `-n` = overwrite existing files
- `-h` = parse to fasta format
- `-e` = input file is in fastq format
- `-i` = converts 'U' to 'T' (RNA to DNA)
- `-j` = remove entries that have entries other than: a,c,g,t,u,n,A,C,G,T,U,N
- `-m` = collapse reads (create one entry but with number indicating its no. of appearances)
- `-k TGGAATTCTCGGGTGCCAAGG` = trim adapter sequence
- `-g` = mismatches (default 1). TamiRNA set this to 2.




miRDeep2.pl - identify novel miRNAs and performs quantification. This script runs all miRDeep2 sub-scripts (excl. mapper.pl as it needs the collapsed fasta from this) which includes quantifier.pl 

quantifier.pl - quantify read counts 

- `-g` = number of mismatches allowed when mapping reads to precursor miRNAs (hairpin.fa). **Default is 1 mismatch**.
