## Methods

### Background Research

This section reviews recent papers that have performed similar analyses to establish miRNA bioinformatic workflows. 

#### <u>Paper 1</u> (2025)

**Title:** A Verified Workflow for MiRNA-Seq Data Processing and Bioinformatics Analysis Using

**Date:** October 2025

**Link**: [here](https://www.jove.com/t/68760/a-verified-workflow-for-mirna-seq-data-processing-bioinformatics)

**Methods:**

##### 1. Preprocess raw reads and perform quality control

- Cutadapt
- fastp
- FastQC

##### 2. Map reads and generate count matrices

**Alignment**
- Align to reference genome
- miRBase (align to mature miRNA sequences)
- Bowtie
1. `-v 0` = no mistakes in entire read
2. `-a --best --strata` = throw away any alignment that had more mismatches than the best

**Counts**
- featureCounts
- HTSeq-count

##### 3. Perform Differential Expression Analysis in R
- DESeq2
1. `dds <- DESeq(dds)` = normalise data
2. `vsd <- vst(dds, blind=FALSE)` = variance-stabilising transformation
- EnhancedVolcano (package)

##### 4. Predict target genes of miRNAs
**Query databases** 
- TargetScan
- miRDB
- miRTarBase

**Automated Prediction**
- multiMiR (package)


##### 5. Conduct functional enrichment analysis
**GO Enrichment**
- clusterProfiler (package)
- org.Hs.eg.db (package)

**KEGG Pathway Enrichment**
- ekegg

##### 6. Construct and visualize miRNA - mRNA interaction network
- Cytoscape


**N.B. Paper includes more in-depth detail on processing steps and parameters, and includes example commands**

**Notes**
- Insert size = 18-30nt for miRNAs
